# `Introduction to pandas Data Structures  - Series`

`A pandas Series is a one-dimensional labeled array that can hold data of a single type, such as integers, floats, strings, or dates. You can think of it as a single column of a spreadsheet or table: it has the values themselves (stored internally as a NumPy array or ExtensionArray) plus an index, which gives each value a label. By default the index is just 0, 1, 2, and so on, but it can be anything, like dates or names, which lets you look up values by label (s["Mon"]) as well as by position (s.iloc[0]). Series support fast vectorized operations (adding two Series aligns them by index automatically), come with a rich set of methods for filtering, aggregating, and transforming data, and serve as the building blocks of a DataFrame, which is essentially a collection of Series sharing the same index.`

```python
a_series = pd.Series([3,4,5,6])
a_series
```

```python
one_d = pd.Series([1,2,3], index=['a', 'b', 'c'])
one_d['a'] = 100
print(one_d)
```

### `A Series is value-mutable but size-immutable in the sense that there's no in-place append. You can change values in place, but you can't grow it the way you'd append to a list.`

```python
one_d[one_d > 2]
```

```python
hundred_randoms = np.random.normal(5, 0.05, 100)

h_series = pd.Series(hundred_randoms)

h_series[h_series > 5]
```

## `Making a pd.series from python dictionary`

```python
data = {"Ohio": 35000, "Texas": 71000, "Oregon": 16000, "Utah": 5000}

series = pd.Series(data)
print(series)
series.to_dict()
```

```python
# california is used as index but has no corresponding value so it will be assigned nan
states = ["California", "Ohio", "Oregon", "Texas"]
data = {"Ohio": 35000, "Texas": 71000, "Oregon": 16000, "Utah": 5000}

obj2 = pd.Series(data, index=states)
print(obj2)
```

### `isna - notna Functions`

```python
pd.isna(obj2)
pd.notna(obj2)
```

```python
# what do you get from this? 
ref_to_index = obj2.index
obj2.index = ["Tenesse", "Ohio", "Oregon", "Texas"]

print(obj2.index is ref_to_index)
```

# `DataFrame`

`A DataFrame represents a rectangular table of data and contains an ordered, named
collection of columns, each of which can be a different value type (numeric, string,
Boolean, etc.). The DataFrame has both a row and column index.`

```python
data = {
    "state": ["Ohio", "Ohio", "Ohio", "Nevada", "Nevada", "Nevada"],
    "year": [2000, 2001, 2002, 2001, 2002, 2003],
    "pop": [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]
}

frame = pd.DataFrame(data)
frame
```

```python
frame.head(2)
```

```python
frame.tail(2)
```

#### `If you specify a sequence of columns, the DataFrame’s columns will be arranged in that order`

```python
df = pd.DataFrame(data, columns=["year", "state", "pop"])
df.head()
```

```python
df.columns
```

### `Retriving specific column returns a series`

```python
df['state']
df.year
```

`frame[column] works for any column name, but frame.column

and does not conflict with any of the method names in DataFrame.
For example, if a column’s name contains whitespace or symbols
other than underscores, it cannot be accessed with the dot attribute
