# `Introduction to pandas Data Structures  - Series`

`A pandas Series is a one-dimensional labeled array that can hold data of a single type, such as integers, floats, strings, or dates. You can think of it as a single column of a spreadsheet or table: it has the values themselves (stored internally as a NumPy array or ExtensionArray) plus an index, which gives each value a label. By default the index is just 0, 1, 2, and so on, but it can be anything, like dates or names, which lets you look up values by label (s["Mon"]) as well as by position (s.iloc[0]). Series support fast vectorized operations (adding two Series aligns them by index automatically), come with a rich set of methods for filtering, aggregating, and transforming data, and serve as the building blocks of a DataFrame, which is essentially a collection of Series sharing the same index.`

```python
a_series = pd.Series([3,4,5,6])
a_series
```

```python
one_d = pd.Series([1,2,3], index=['a', 'b', 'c'])
one_d['a'] = 100
print(one_d)
```

### `A Series is value-mutable but size-immutable in the sense that there's no in-place append. You can change values in place, but you can't grow it the way you'd append to a list.`

```python
one_d[one_d > 2]
```

```python
hundred_randoms = np.random.normal(5, 0.05, 100)

h_series = pd.Series(hundred_randoms)

h_series[h_series > 5]
```

## Mean & Median & Mode

```python
def mean(values):
    
    if len(values) == 0:
        raise ValueError("mean() needs at least one value")
    return sum(values) / len(values)



def median(values):
    
    if len(values) == 0:
        raise ValueError("median() needs at least one value")
    s = sorted(values)          
    n = len(s)
    mid = n // 2
    if n % 2 == 1:              
        return s[mid]
    return (s[mid - 1] + s[mid]) / 2   



def mode(values):
    if len(values) == 0:
        raise ValueError("mode() needs at least one value")
    counts = {}
    for v in values:
        counts[v] = counts.get(v, 0) + 1
    highest = max(counts.values())
    return [v for v, c in counts.items() if c == highest]
```

## Titanic Dataset: Feature Explanations

- `PassengerId`: a unique ID number for each passenger (just a row identifier, not useful for prediction)
- `Survived`: whether the passenger survived, 0 = no, 1 = yes (the usual prediction target; missing in Kaggle's test file)
- `Pclass`: ticket class, 1 = first (upper), 2 = second (middle), 3 = third (lower), a proxy for social and economic status
- `Name`: the passenger's full name, including a title such as Mr, Mrs, Miss, or Master (the title can be extracted as a feature)
- `Sex`: the passenger's sex, male or female
- `Age`: age in years (fractional if under 1; has many missing values)
- `SibSp`: number of siblings or spouses aboard with the passenger
- `Parch`: number of parents or children aboard with the passenger
- `Ticket`: the ticket number (several passengers can share one ticket)
- `Fare`: the price paid for the ticket, in pounds (strongly right-skewed, with a few very expensive tickets)
- `Cabin`: the cabin number, whose first letter shows the deck (mostly missing)
- `Embarked`: port of embarkation, C = Cherbourg, Q = Queenstown, S = Southampton



1) how much of the cabin feature is missing?
