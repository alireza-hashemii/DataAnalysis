## `Slicing and Indexing`

Random data about 10 people's age,systolic blood pressure, resting heart beat

```python
# age (years), systolic blood pressure (mmHg), resting heart rate (bpm)
patients = np.array([
    [72, 118, 68],
    [34, 122, 72],
    [41, 128, 75],
    [29, 115, 64],
    [52, 135, 80],
    [60, 142, 78],
    [47, 130, 74],
    [38, 124, 70],
    [66, 148, 82],
    [31, 120, 66],
])

print('Average Blood Pressure is',patients[:, 1].mean())
```

```python
# separating those with bp higher than 130
over_130 = patients[patients[:, 1] > 130]
print(over_130)
```

```python
# separating those under age of 50
age_under_50 = patients[patients[:,0] < 50]
age_under_50
```

```python
# display only the resting heart beat feature
print(patients[:, 2])
```

```python
# sum of every row
patients.sum(axis=1)

# sum of every column
patients.sum(axis=0)
```

```python
# average age of people with blood pressure higher than 140
condition_result = patients[patients[:, 1] < 140] 
condition_result[:, 0].mean(axis=0)
```

```python
# sort based on blood pressure
order = patients[:, 1].argsort(axis=0)
patients_sordet = patients[order]
patients_sordet
```

```python
# which patient has the lowest age (is the youngest)
patients[patients[:, 0].argmin(axis=0)]
```

## `Boolean Indexing`

```python
# columns: hours_studied, exam_score, absences
students = np.array([
    [ 2, 55, 6],
    [ 5, 72, 2],
    [ 8, 88, 0],
    [ 1, 48, 8],
    [ 6, 79, 3],
    [ 4, 65, 4],
    [ 9, 93, 1],
    [ 3, 58, 5],
    [ 7, 84, 2],
    [10, 91, 0],
])

hours = students[:, 0]
scores = students[:, 1]
absences = students[:, 2]
```

```python
mask = scores >= 80
print(mask)
```

Combine conditions with & (and), | (or), ~ (not). Each condition needs its own parentheses

```python
students[(hours >= 5) & (absences <= 2)]
```

1. Which exam scores are 80 or higher?<br>
2. How many students studied at least 5 hours and had at most 2 absences?<br>
3. Show the rows of students with a score below 60 or at least 6 absences.<br>
4. What is the mean score of students who did not score 80 or higher? (Use ~.)<br>
5. What is the mean number of absences among students scoring 80 or higher?<br>
6. Bonus: how many students scored above the class average?<br>

```python
# answer of question number 1 
mask = students[:, 1] > 80
students[mask]
```

```python
# answer of question number 2 
students[(hours >= 5) & (absences <= 2)]
```

```python
# answer of question number 3
students[(scores < 60) | (absences >= 6)]
```

```python
# answer of question number 4 
mask = students[:, 1] < 80
selected_rows = students[mask]
selected_rows[:, 1].mean()
```

```python
# answer of question number 5 
mask = students[:, 1] > 80
over_80_stds = students[mask]
avg_abc = over_80_stds[:, 2].mean()
avg_abc
```

```python
# answer of question number 6
avg = students[:, 1].mean()
mask = students[:, 1] > avg
better_half = students[mask]

better_half
```
