## `Boolean Indexing`

````python
# columns: hours_studied, exam_score, absences
students = np.array([
    [ 2, 55, 6],
    [ 5, 72, 2],
    [ 8, 88, 0],
    [ 1, 48, 8],
    [ 6, 79, 3],
    [ 4, 65, 4],
    [ 9, 93, 1],
    [ 3, 58, 5],
    [ 7, 84, 2],
    [10, 91, 0],
])

hours = students[:, 0]
scores = students[:, 1]
absences = students[:, 2]
````

````python
mask = scores >= 80
print(mask)
````

Combine conditions with & (and), | (or), ~ (not). Each condition needs its own parentheses

````python
students[(hours >= 5) & (absences <= 2)]
````

# `NumPy Universal Functions (Element-wise)`

| Function | Description | Example | Result |
|---|---|---|---|
| `np.abs(x)` | Absolute value (removes the negative sign) | `np.abs(-5)` | `5` |
| `np.sqrt(x)` | Square root | `np.sqrt(16)` | `4.0` |
| `np.square(x)` | Squares each element (x²) | `np.square(3)` | `9` |
| `np.sign(x)` | Returns -1 for negative, 0 for zero, 1 for positive | `np.sign(-7)` | `-1` |
| `np.isnan(x)` | `True` where the value is NaN (not a number) | `np.isnan(np.nan)` | `True` |
| `np.isinf(x)` | `True` where the value is positive or negative infinity | `np.isinf(np.inf)` | `True` |
| `np.ceil(x)` | Rounds up to the nearest integer | `np.ceil(2.1)` | `3.0` |
| `np.floor(x)` | Rounds down to the nearest integer | `np.floor(2.9)` | `2.0` |
| `np.log(x)` | Natural logarithm (base e) | `np.log(np.e)` | `1.0` |
| `np.exp(x)` | Exponential (e raised to x) | `np.exp(1)` | `2.718...` |

## Notes

- All of these work **element-wise**, so they apply to every entry of an array:


- `np.log` and `np.exp` are inverses: `np.log(np.exp(x))` gives back `x`.
- `np.sqrt` of a negative number returns `nan` (with a warning), and `np.log(0)` returns `-inf`. This is where `np.isnan` and `np.isinf` come in handy.
- `np.isnan` and `np.isinf` return **boolean arrays**, which are useful for filtering bad data:



# `Expressing Conditional Logic as Array Operations`

````python
# pythonic approach

xarr = np.array([1.1, 1.2, 1.3, 1.4, 1.5])
yarr = np.array([2.1, 2.2, 2.3, 2.4, 2.5])
cond = np.array([True, False, True, True, False])


result = np.array([(x if c else y) for x, y, c in zip(xarr, yarr, cond)])
result
````

````python
# an alternativ in numpy

result = np.where(cond, xarr, yarr)
result
````

# NumPy Aggregation Functions

| Function | Description | Example | Result |
|---|---|---|---|
| `np.sum(x)` | Sum of all elements, or along an axis; zero-length arrays give 0 | `np.sum([1, 2, 3])` | `6` |
| `np.mean(x)` | Arithmetic mean; returns `NaN` on zero-length arrays | `np.mean([1, 2, 3])` | `2.0` |
| `np.std(x)` | Standard deviation | `np.std([1, 2, 3])` | `0.816...` |
| `np.var(x)` | Variance | `np.var([1, 2, 3])` | `0.666...` |
| `np.min(x)` | Minimum value | `np.min([4, 1, 7])` | `1` |
| `np.max(x)` | Maximum value | `np.max([4, 1, 7])` | `7` |
| `np.argmin(x)` | Index of the minimum element | `np.argmin([4, 1, 7])` | `1` |
| `np.argmax(x)` | Index of the maximum element | `np.argmax([4, 1, 7])` | `2` |
| `np.cumsum(x)` | Cumulative sum, starting from 0 | `np.cumsum([1, 2, 3])` | `[1, 3, 6]` |
| `np.cumprod(x)` | Cumulative product, starting from 1 | `np.cumprod([1, 2, 3])` | `[1, 2, 6]` |

## Notes

- Most of these accept an `axis` argument. With a 2D array, `axis=0` works down the columns and `axis=1` works across the rows:

```python
  A = np.array([[1, 2, 3],
                [4, 5, 6]])
  A.sum()          # 21
  A.sum(axis=0)    # [5, 7, 9]   (one result per column)
  A.sum(axis=1)    # [6, 15]     (one result per row)
```

- They are also available as array methods: `A.sum()`, `A.mean()`, `A.argmax()`, etc.
- `std` and `var` use the population formula by default (`ddof=0`). For the sample version, use `np.std(x, ddof=1)`.
- `argmin` and `argmax` on a 2D array without `axis` return an index into the flattened array.
